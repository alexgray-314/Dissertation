```python
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator

# Example data
categories = ["Visibility","Role-Expressiveness","Delayed Commitment","Closeness of mapping","Abstraction","Consistency","Progressive Evaluation", "Provisionality",
"Hard mental operations","Explicit dependencies","Error-robust","Secondary notation"]
values = [0.438,	0.125,	0.625,	0.5,	0.375,	0.875,	0.1875,	0.875,	0.0625,	0.375,	0.125,	0.75]

# Create horizontal bar chart
plt.figure(figsize=(10, 5))
plt.barh(categories, values)

# Tick labels
plt.xticks([0, 0.5, 1], ["Neutral", "Agree", "Strongly Agree"])
ax = plt.gca()
ax.xaxis.set_minor_locator(MultipleLocator(0.05))  # grid every 0.1

# Gridlines
plt.grid(axis='x', linestyle='-', alpha=0.3, which='minor')
plt.grid(axis='x', linestyle='-', alpha=0.8, which='major')

# Labels and title
plt.xlabel("Average likert response across all participants", fontweight='bold')
plt.ylabel("Cognitive dimension", fontweight='bold')
plt.title("User Feedback on Language Comprehensibility and Conciseness\n Across Cognitive Dimensions", fontweight='bold')

# Optional: make layout cleaner
plt.tight_layout()

# Show plot
plt.show()
```
